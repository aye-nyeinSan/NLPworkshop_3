```python
import nltk
import pandas as pd
import re

#load the dataset
df = pd.read_csv("Datasets/spam.csv",encoding='latin-1')
print("loaded dataset!")
df
```

```python
# Drop the specified columns
df.drop(columns=["Unnamed: 2", "Unnamed: 3", "Unnamed: 4"], inplace=True)

# Display the updated DataFrame
print(df)
```

```python
#remove whitespaces and convert to lowercase 
df['v2'] = df['v2'].apply(lambda x: x.lower())
df['v2'] = df['v2'].apply(lambda x: re.sub(r'\s+', ' ', x))
df['v2']
```

```python
#remove anything that is not English
df['v2'] = df['v2'].apply(lambda x: re.sub(r'[^a-zA-Z]', ' ', x))
df['v2']
```

```python
if 'v1' in df.columns and 'v2' in df.columns:
    df['label'] = df['v1']
    df['text'] = df['v2']
else:
    print("Column 'v1' and 'v2' does not exist in the DataFrame.")
    df.drop(columns=['v1', 'v2'], inplace=True)

df = df[['label', 'text']]
df['length'] = df['text'].apply(lambda x: len(x.split()))
    
df
```

```python

import nltk

# Ensure nltk resources are downloaded
nltk.download('words')
nltk.download('punkt')
words = set(nltk.corpus.words.words())


# Assuming `words` is a set of valid vocabulary words
# Example: words = set(["word1", "word2", "example", ...])
df['text2'] = df['text'].astype(str).apply(
    lambda x: " ".join(
        w for w in nltk.wordpunct_tokenize(x) 
        if w.lower() in words or not w.isalpha()
    )
)

df
```

```python
#labelEncoding the target variable 
!pip3 install scikit-learn
from sklearn.preprocessing import LabelEncoder
#label encoding the target variable(ham,spam) into numerical data

df['label_transform']= LabelEncoder().fit_transform(df['label'])
df['label_transform'].value_counts()
#0 is ham and 1 is spam
```

```python
#count the count of unique words changing text data  to numerical data
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(stop_words=None,token_pattern=r'\b\w+\b')
matrix = vectorizer.fit_transform(df['text2'])

# Convert to array and inspect the result
print(matrix.toarray())  # This should now contain non-zero values
print("Total unique words:", len(vectorizer.get_feature_names_out()))  # Total unique words
```

```python
matrix_array = pd.DataFrame(matrix.toarray(),columns= vectorizer.get_feature_names_out())
matrix_array
```

```python
#Sorting the top 10 words
import pandas as pd

# Calculate the sum for each feature (column)
word_sums = matrix_array.sum(axis=0)

# Sort the words by their total frequency in descending order
top_words = word_sums.sort_values(ascending=False).head(10)

print(top_words)
```

```python
#first 5 rows of the transformed dataset
print ("Top 5 rows of the transformed dataset")
matrix_array.head(5)

```

```python
#Bottom 5 rows of the transformed dataset
print ("Bottom 5 rows of the transformed dataset")
matrix_array.tail(5)
```

```python
#KDD (Knowledge Discovery in database ) process 
#split the BOW data into training and testing data

from sklearn.model_selection import train_test_split
X= matrix_array #features=bag of words matrix
Y= df['label_transform'] #target variable

# Split data into training (80%) and testing (20%)
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Train a logistic regression model
classifier = LogisticRegression()
classifier.fit(X_train, Y_train)
y_predict = classifier.predict(X_test)
```

```python
#measure the accuracy of the model
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test, y_predict)
print("Model accuracy: {:.2f}%".format(accuracy * 100))
```

Precision, Recall, and F1-Score: These metrics provide additional insights into the performance of your classifier.

Accuracy: Indicates how well the model performed overall.

Precision: Shows how many predicted spam messages were actually spam.

Recall: Reflects how many actual spam messages were correctly identified.

F1-Score: A harmonic mean of precision and recall, useful for imbalanced datasets.

```python
from sklearn.metrics import classification_report

report = classification_report(Y_test, y_predict, target_names=['ham', 'spam'])
print("Classification Report:\n", report)
```

### Using TF-IDF technique  

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Use the same vectorizer instance
tfid_vectorizer = TfidfVectorizer()
X = tfid_vectorizer.fit_transform(df['text2'])  # TF-IDF sparse matrix
X_df = pd.DataFrame(X.toarray(), columns=tfid_vectorizer.get_feature_names_out())
X_df
```

### Feature Selection:

Removes features (words) with low variance (below the threshold of 0.1), which are likely unimportant for classification.

```python
# Convert TF-IDF sparse matrix to DataFrame for better analysis
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0.0)
X_selected = selector.fit_transform(X_df)
X_df.shape[1], X_selected.shape[1]
```

```python
# Number of features removed
features_removed = X_df.shape[1] - X_selected.shape[1]
print(f"Number of features removed: {features_removed}")
```

###  Stratified hold-out with 70:30 ratio

```python
from sklearn.model_selection import train_test_split

# Sort the dataset by label to simulate a non-shuffled stratified split
df_sorted = df.sort_values(by='label_transform')

# Extract features and target
X = X_selected
y = df_sorted['label_transform']

# Perform a standard train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1234, stratify=None, shuffle=False
)

```
